## 1. Environment and Imports

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from data.clean_dataset import generate_and_clean_data
from models.academic_predictor import AcademicPredictor
from vision.attendance_detector import AttendanceDetector
from nlp.feedback_analyzer import FeedbackAnalyzer
from visualizer.dashboard import generate_analytics_dashboard

print("Environment initialized successfully!")
```

## 2. Data Preparation & RegEx Cleaning

## 3. Machine Learning: Performance & Risk Prediction

```python
predictor = AcademicPredictor()
metrics = predictor.train("data/student_records.csv")

print(f"Decision Tree Accuracy: {metrics['classification_accuracy']:.2%}")
print(f"Linear Regression R2   : {metrics['regression_r2']:.3f}")
print("\nClassification Report:\n", metrics['classification_report'])
print("Feature Importances:", metrics['feature_importances'])
```

## 4. Computer Vision: Face Detection & Classroom Headcount

```python
detector = AttendanceDetector()
demo_img, count = detector.run_on_synthetic_classroom("data/classroom_demo.jpg")
print(f"Automated Attendance Count: {count} students present")

Image.open(demo_img)
```

## 5. NLP: Sentiment Analysis & Feedback WordCloud

```python
analyzer = FeedbackAnalyzer()
_, nlp_summary = analyzer.analyze_dataset_feedback("data/student_records.csv")
wc_path, top_words = analyzer.generate_wordcloud("data/student_records.csv", "data/feedback_wordcloud.png")

print("Sentiment Summary:", nlp_summary)
print("Top Theme Keywords:", top_words)
Image.open(wc_path)
```
