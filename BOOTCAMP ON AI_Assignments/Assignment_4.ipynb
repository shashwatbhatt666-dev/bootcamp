# OpenCV Image Editing Application

## Aim

To develop an image editing application using **OpenCV-Python** that demonstrates various image processing and editing techniques such as resizing, brightness and contrast adjustment, grayscale conversion, blurring, sharpening, edge detection, rotation, negative effects, watermarking, and image saving.

## Introduction

**OpenCV (Open Source Computer Vision Library)** is a powerful Python library used for image processing, computer vision, and image analysis.

In this program, an input image is loaded and several image editing operations are performed using OpenCV. The program also creates a **before-and-after comparison** and saves the processed images for future use.

## Objectives

The program demonstrates the following image editing operations:

* Loading an image
* Resizing an image
* Adjusting brightness
* Adjusting contrast
* Converting an image to grayscale
* Applying Gaussian blur
* Sharpening an image
* Detecting edges using the Canny algorithm
* Creating a negative/inverted image
* Rotating an image
* Adding a text watermark
* Displaying edited images
* Saving processed images

## Libraries Used

### OpenCV

`cv2` is used for image reading, processing, transformation, displaying, and saving.

### NumPy

`numpy` is used to create and manipulate the sharpening kernel and perform array-based image operations.

## Image Editing Operations

### 1. Image Loading

The input image is loaded using:

```python
cv2.imread()
```

The program also checks whether the image was loaded successfully.

### 2. Image Resizing

The image is resized to a fixed resolution of **800 × 600 pixels** using:

```python
cv2.resize()
```

### 3. Brightness Adjustment

Brightness is increased using:

```python
cv2.convertScaleAbs()
```

The `beta` parameter controls the brightness of the image.

### 4. Contrast Adjustment

Contrast is enhanced using the `alpha` parameter of `cv2.convertScaleAbs()`.

A value greater than `1` increases the contrast.

### 5. Grayscale Conversion

The colored BGR image is converted into a grayscale image using:

```python
cv2.cvtColor()
```

with:

```python
cv2.COLOR_BGR2GRAY
```

Grayscale images contain intensity information instead of three color channels.

### 6. Gaussian Blur

Gaussian blur is applied using:

```python
cv2.GaussianBlur()
```

Blurring can be used for noise reduction and image smoothing.

### 7. Image Sharpening

A sharpening kernel is created using NumPy and applied using:

```python
cv2.filter2D()
```

This enhances edges and fine details in the image.

### 8. Edge Detection

The **Canny Edge Detection** algorithm is used to identify prominent edges:

```python
cv2.Canny()
```

This is commonly used in computer vision applications for detecting boundaries and shapes.

### 9. Negative Effect

The colors of the image are inverted using:

```python
cv2.bitwise_not()
```

This produces a negative or inverted version of the original image.

### 10. Image Rotation

The image is rotated by **90 degrees clockwise** using:

```python
cv2.rotate()
```

### 11. Adding a Watermark

Text is added to the image using:

```python
cv2.putText()
```

The watermark helps demonstrate text rendering directly onto an image.

### 12. Before and After Comparison

The original and edited images are combined horizontally using:

```python
np.hstack()
```

This makes it easy to visually compare the original image with the edited version.

### 13. Saving Images

The processed images are saved using:

```python
cv2.imwrite()
```

The program generates separate files for the grayscale, blurred, sharpened, edge-detected, negative, rotated, and edited images.

## Expected Output

The program displays multiple windows containing:

1. Original vs. edited image
2. Grayscale image
3. Blurred image
4. Sharpened image
5. Edge-detected image
6. Negative image
7. Rotated image
8. Watermarked image

The processed images are also saved to the current working directory.

## Conclusion

This program demonstrates the use of **OpenCV-Python for practical image editing and processing**. By combining multiple operations such as enhancement, filtering, transformation, edge detection, and watermarking, the program provides a basic foundation for developing more advanced computer vision and image-editing applications.


````python
import cv2
import numpy as np

# ============================================================
#        OPENCV IMAGE EDITING APPLICATION
# ============================================================

# Load image
image_path = "1.jpg"
img = cv2.imread(image_path)

if img is None:
    print("Error: Could not load image.")
    print("Check the image path.")
    exit()

# Make a copy for editing
edited = img.copy()

print("\n" + "=" * 55)
print("           OPENCV IMAGE EDITING APPLICATION")
print("=" * 55)

print("\nImage loaded successfully!")
print("Original Size:", img.shape[1], "x", img.shape[0])


# ------------------------------------------------------------
# 1. RESIZE IMAGE
# ------------------------------------------------------------

edited = cv2.resize(edited, (800, 600))

print("\nImage resized to 800 x 600")


# ------------------------------------------------------------
# 2. BRIGHTNESS ADJUSTMENT
# ------------------------------------------------------------

brightness = 30

bright = cv2.convertScaleAbs(
    edited,
    alpha=1.0,
    beta=brightness
)


# ------------------------------------------------------------
# 3. CONTRAST ADJUSTMENT
# ------------------------------------------------------------

contrast = 1.3

contrast_img = cv2.convertScaleAbs(
    bright,
    alpha=contrast,
    beta=0
)


# ------------------------------------------------------------
# 4. GRAYSCALE CONVERSION
# ------------------------------------------------------------

gray = cv2.cvtColor(contrast_img, cv2.COLOR_BGR2GRAY)

# Convert back to BGR so it can be displayed consistently
gray_bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)


# ------------------------------------------------------------
# 5. GAUSSIAN BLUR
# ------------------------------------------------------------

blurred = cv2.GaussianBlur(
    contrast_img,
    (9, 9),
    0
)


# ------------------------------------------------------------
# 6. IMAGE SHARPENING
# ------------------------------------------------------------

kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])

sharpened = cv2.filter2D(
    contrast_img,
    -1,
    kernel
)


# ------------------------------------------------------------
# 7. EDGE DETECTION
# ------------------------------------------------------------

edges = cv2.Canny(
    gray,
    100,
    200
)

edges_bgr = cv2.cvtColor(
    edges,
    cv2.COLOR_GRAY2BGR
)


# ------------------------------------------------------------
# 8. NEGATIVE / INVERT EFFECT
# ------------------------------------------------------------

negative = cv2.bitwise_not(
    contrast_img
)


# ------------------------------------------------------------
# 9. ROTATE IMAGE
# ------------------------------------------------------------

rotated = cv2.rotate(
    contrast_img,
    cv2.ROTATE_90_CLOCKWISE
)


# ------------------------------------------------------------
# 10. ADD WATERMARK / TEXT
# ------------------------------------------------------------

watermarked = contrast_img.copy()

cv2.putText(
    watermarked,
    "Edited with OpenCV",
    (30, 560),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (255, 255, 255),
    2,
    cv2.LINE_AA
)


# ------------------------------------------------------------
# 11. CREATE BEFORE / AFTER COMPARISON
# ------------------------------------------------------------

original_display = cv2.resize(
    img,
    (800, 600)
)

comparison = np.hstack(
    (original_display, watermarked)
)

cv2.putText(
    comparison,
    "ORIGINAL",
    (250, 50),
    cv2.FONT_HERSHEY_SIMPLEX,
    1.2,
    (0, 255, 0),
    3
)

cv2.putText(
    comparison,
    "EDITED",
    (1050, 50),
    cv2.FONT_HERSHEY_SIMPLEX,
    1.2,
    (0, 255, 0),
    3
)


# ------------------------------------------------------------
# 12. DISPLAY RESULTS
# ------------------------------------------------------------

cv2.imshow(
    "Before vs After",
    comparison
)

cv2.imshow(
    "Grayscale",
    gray_bgr
)

cv2.imshow(
    "Blurred",
    blurred
)

cv2.imshow(
    "Sharpened",
    sharpened
)

cv2.imshow(
    "Edge Detection",
    edges_bgr
)

cv2.imshow(
    "Negative Effect",
    negative
)

cv2.imshow(
    "Rotated Image",
    rotated
)

cv2.imshow(
    "Watermarked Image",
    watermarked
)


# ------------------------------------------------------------
# 13. SAVE EDITED IMAGES
# ------------------------------------------------------------

cv2.imwrite(
    "edited_image.jpg",
    watermarked
)

cv2.imwrite(
    "grayscale_image.jpg",
    gray
)

cv2.imwrite(
    "blurred_image.jpg",
    blurred
)

cv2.imwrite(
    "sharpened_image.jpg",
    sharpened
)

cv2.imwrite(
    "edge_detection.jpg",
    edges
)

cv2.imwrite(
    "negative_image.jpg",
    negative
)

cv2.imwrite(
    "rotated_image.jpg",
    rotated
)

cv2.imwrite(
    "before_after.jpg",
    comparison
)


print("\n" + "=" * 55)
print("             IMAGE EDITING COMPLETED")
print("=" * 55)

print("\nFiles saved successfully:")
print("1. edited_image.jpg")
print("2. grayscale_image.jpg")
print("3. blurred_image.jpg")
print("4. sharpened_image.jpg")
print("5. edge_detection.jpg")
print("6. negative_image.jpg")
print("7. rotated_image.jpg")
print("8. before_after.jpg")

print("\nPress any key on an image window to exit.")

cv2.waitKey(0)
cv2.destroyAllWindows()

````
