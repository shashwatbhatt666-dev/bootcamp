# NumPy, Pandas and Matplotlib Experiment

This experiment demonstrates the use of **NumPy, Pandas, and Matplotlib** for data generation, data handling, data cleaning, and visualization.

### Topics Covered

- Creating 2D random data using **NumPy**
- Creating and working with a **Pandas DataFrame**
- Adding column names and student names
- Saving data into a **CSV file**
- Cleaning messy text data using **Regular Expressions (RegEx)**
- Removing spaces and special characters from strings
- Creating a **sine wave** using NumPy
- Visualizing data using **Matplotlib**

### Objective

To understand how NumPy can be used for numerical data generation, Pandas can be used for data handling and cleaning, and Matplotlib can be used for data visualization.

### Conclusion

This experiment demonstrates a basic data analysis workflow: **generate data → create a dataset → save data → clean data → visualize data** using Python libraries.

```python
# ==========================================
# NUMPY, PANDAS AND MATPLOTLIB EXPERIMENT
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------
# 1. Create 2D random data using NumPy
# ------------------------------------------

np.random.seed(42)

data = np.random.randint(10, 101, size=(10, 3))

print("2D Random Data:")
print(data)


# ------------------------------------------
# 2. Create Pandas DataFrame
# ------------------------------------------

df = pd.DataFrame(
    data,
    columns=["Maths", "Physics", "Programming"]
)

df.insert(0, "Name", [
    "Rahul", "Aman", "Rohan", "Vivek", "Arjun",
    "Karan", "Aditya", "Sahil", "Vishal", "Ankit"
])

print("\nStudent Dataset:")
print(df)


# ------------------------------------------
# 3. Save DataFrame as CSV
# ------------------------------------------

df.to_csv("student_data.csv", index=False)

print("\nCSV file created successfully!")


# ------------------------------------------
# 4. Data Cleaning using RegEx
# ------------------------------------------

messy_data = {
    "Name": [
        "  Rahul...Sharma ",
        "Aman///Kumar",
        "..Rohan@Singh..",
        " Vivek###Patel ",
        "Arjun$$Gupta"
    ],
    "Marks": [85, 76, 91, 68, 88]
}

clean_df = pd.DataFrame(messy_data)

print("\nBefore Cleaning:")
print(clean_df)

# Remove leading and trailing spaces
clean_df["Name"] = clean_df["Name"].str.strip()

# Replace special characters with spaces
clean_df["Name"] = clean_df["Name"].str.replace(
    r"[^A-Za-z ]", " ", regex=True
)

# Remove extra spaces
clean_df["Name"] = clean_df["Name"].str.replace(
    r"\s+", " ", regex=True
)

# Remove spaces at beginning/end
clean_df["Name"] = clean_df["Name"].str.strip()

print("\nAfter Cleaning:")
print(clean_df)


# ------------------------------------------
# 5. Draw a Sine Wave
# ------------------------------------------

x = np.linspace(0, 2 * np.pi, 500)

y = np.sin(x)

plt.figure(figsize=(8, 5))

plt.plot(x, y, linewidth=2)

plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")

plt.grid(True)

plt.show()
```
