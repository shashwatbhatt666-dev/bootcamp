### Aim
To create a list of random numbers and print all the even numbers from the list.

### Explanation
The `random` module is used to generate random numbers between 1 and 100. 
A `for` loop checks each number using the condition `num % 2 == 0`. 
If the remainder is 0, the number is even and is printed.

```python
import random

# Create a list of 10 random numbers
numbers = [random.randint(1, 100) for i in range(10)]

print("List of numbers:", numbers)

# Print all even numbers
print("Even numbers:")

for num in numbers:
    if num % 2 == 0:
        print(num, end=" ")
```

### Aim
To create a string, print it in reverse order, and reverse the order of its words.

### Explanation
String slicing `[::-1]` is used to reverse the complete string. 
The `split()` function separates the string into words, and `[::-1]` reverses their order. 
Finally, `join()` combines the reversed words into a single string.

```python
# Create a string
text = "Python is easy to learn"

# Reverse the complete string
print("String in reverse order:")
print(text[::-1])

# Reverse the order of words
words = text.split()

print("Words in reverse order:")
print(" ".join(words[::-1]))
```
