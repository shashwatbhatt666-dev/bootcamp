# Day 2: NumPy, Pandas and Matplotlib

## 1. NumPy Operations: Sports Data Analysis

```python
import numpy as np
players = np.array(["Player 1", "Player 2", "Player 3", "Player 4", "Player 5"])
matches = np.array(["Match 1", "Match 2", "Match 3"])
np.random.seed(42)
runs = np.random.randint(10, 100, size=(len(players), len(matches)))
print("Players:", players)
print("Matches:", matches)
print("Runs:\n", runs)
total_runs_by_player = runs.sum(axis=1)
for player, total in zip(players, total_runs_by_player):
    print(f"{player}: {total} runs")
average_runs_per_match = runs.mean(axis=0)
for match, avg in zip(matches, average_runs_per_match):
    print(f"{match}: {avg:.2f} runs")
most_runs_index = np.argmax(total_runs_by_player)
print(f"Top scorer: {players[most_runs_index]} with {total_runs_by_player[most_runs_index]} runs")
std_dev_by_player = runs.std(axis=1)
for player, std in zip(players, std_dev_by_player):
    print(f"{player} standard deviation: {std:.2f}")
```

## 2. Statistical Analysis: Player Scores

```python
players = np.array(["Rahul", "Aman", "Rohan", "Vivek", "Arjun"])
scores = np.random.randint(1, 101, size=(5, 3))
print("Players:\n", players)
print("Scores:\n", scores)
print("Row-wise Sum:", np.sum(scores, axis=1))
print("Column-wise Sum:", np.sum(scores, axis=0))
print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
values, frequency = np.unique(scores, return_counts=True)
mode = values[np.argmax(frequency)]
print("Mode:", mode)
print("Standard Deviation:", np.std(scores))
```

## 3. 3D NumPy Array: Olympic Medals Analysis

```python
countries = np.array(['USA', 'China', 'India', 'Germany', 'Australia'])
years = np.array([2012, 2016, 2020, 2024])
np.random.seed(42)
medals = np.random.randint(10, 100, size=(len(countries), len(years), 3))
print("Countries:", countries)
print("Years:", years)
total_medals_by_country = medals.sum(axis=2)
for country, total in zip(countries, total_medals_by_country):
    print(f"{country}: {total}")
medals_2020 = total_medals_by_country[:, 2]
most_medals_index = np.argmax(medals_2020)
print(f"Top country in 2020: {countries[most_medals_index]} ({medals_2020[most_medals_index]} medals)")
average_medals_by_year = total_medals_by_country.mean(axis=0)
for year, avg in zip(years, average_medals_by_year):
    print(f"{year} average: {avg:.2f}")
std_dev_by_year = total_medals_by_country.std(axis=0)
for year, std in zip(years, std_dev_by_year):
    print(f"{year} std dev: {std:.2f}")
```

## 4. Matplotlib: Total Runs and Olympic Medals Plots

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.bar(players, total_runs_by_player, color='skyblue')
plt.title("Total Runs Scored by Each Player")
plt.xlabel("Players")
plt.ylabel("Total Runs")
plt.show()
plt.figure(figsize=(10, 6))
for i, country in enumerate(countries):
    plt.plot(years, total_medals_by_country[i], label=country, marker='o')
plt.title('Total Olympic Medals by Country Over Years')
plt.xlabel('Year')
plt.ylabel('Total Medals')
plt.legend()
plt.grid(True)
plt.show()
```

## 5. Harmonic Motion Visualization

```python
time = np.linspace(0, 10, 500)
amplitude1 = np.sin(2 * np.pi * time)
amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), dpi=100)
ax1.plot(time, amplitude1, color="crimson", linewidth=2, label="Undamped Harmonic Motion")
ax1.set_title("Harmonic Response of Structure")
ax1.set_xlabel("Time (seconds)")
ax1.set_ylabel("Displacement (mm)")
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right")
ax2.plot(time, amplitude2, color="navy", linewidth=2, label="Damped Harmonic Motion")
ax2.set_title("Damped Harmonic Response of Structure")
ax2.set_xlabel("Time (seconds)")
ax2.set_ylabel("Displacement (mm)")
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right")
plt.tight_layout()
plt.show()
```

## 6. Animated Sine Wave

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
x = np.linspace(0, 2 * np.pi, 200)
fig, ax = plt.subplots()
line, = ax.plot(x, np.sin(x), lw=3)
ax.set_ylim(-1.2, 1.2)
def update(phase):
    line.set_ydata(np.sin(x + phase))
    return line,
ani = FuncAnimation(fig, update, frames=np.linspace(0, np.pi / 2, 60), interval=40, blit=True)
HTML(ani.to_jshtml())
```

## 8. Pandas: Indian Singers Dataset Analysis

```python
import pandas as pd
data = {
    'Singer': ['Arijit Singh', 'Shreya Ghoshal', 'Sonu Nigam', 'Lata Mangeshkar', 'Kishore Kumar'],
    'Years_Active': [15, 20, 30, 70, 40],
    'Number_of_Songs': [500, 700, 600, 1000, 800],
    'Awards_Won': [50, 60, 40, 100, 80]
}
singers_df = pd.DataFrame(data)
singers_df.info()
print(singers_df.describe())
print("Max songs:", singers_df['Number_of_Songs'].max())
print("Min songs:", singers_df['Number_of_Songs'].min())
print(f"Average awards won: {singers_df['Awards_Won'].mean():.2f}")
print("Median years active:", singers_df['Years_Active'].median())
sorted_singers = singers_df.sort_values(by='Awards_Won', ascending=False)
print("Sorted by awards:\n", sorted_singers)
singers_with_many_awards = singers_df[singers_df['Awards_Won'] > 50]
print("Singers with > 50 awards:\n", singers_with_many_awards)
plt.figure(figsize=(8, 5))
plt.bar(singers_df['Singer'], singers_df['Number_of_Songs'], color='lightblue')
plt.title('Number of Songs by Indian Singers')
plt.xlabel('Singer')
plt.ylabel('Number of Songs')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

## 9. Kumbh Mela Visitor Analysis

```python
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5']
visitors = np.random.randint(50000, 200000, size=5)
cities = ['Prayagraj', 'Varanasi', 'Lucknow', 'Kanpur', 'Agra']
participants = np.random.randint(3000, 15000, size=5)
df_visitors = pd.DataFrame({'Day': days, 'Visitors': visitors})
df_participants = pd.DataFrame({'City': cities, 'Participants': participants})
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(df_visitors['Day'], df_visitors['Visitors'], marker='s', color='green')
plt.title('Daily Visitors')
plt.xlabel('Days')
plt.ylabel('Visitors')
plt.subplot(1, 2, 2)
plt.bar(df_participants['City'], df_participants['Participants'], color='purple')
plt.title('City-wise Participants')
plt.xlabel('Cities')
plt.ylabel('Participants')
plt.tight_layout()
plt.show()
plt.figure(figsize=(6, 6))
plt.pie(df_participants['Participants'], labels=df_participants['City'], autopct='%1.1f%%', colors=plt.cm.Paired.colors)
plt.title('Participant Distribution by City')
plt.show()
```
