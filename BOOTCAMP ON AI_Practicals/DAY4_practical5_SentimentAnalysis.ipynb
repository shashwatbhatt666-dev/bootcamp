**---------Real-Time Sentiment & Emotion Analyzer----------**

Instead of loading massive deep learning models, this uses TextBlob (built on NLTK). 
It measures Polarity (how positive/negative text is) and Subjectivity (how opinionated vs. factual text is) instantly.

The Hook: Live feedback as we type different reviews or customer complaints into the terminal.

Core Concept: Tokenization, Lexicon-based Sentiment Scoring, and Subjectivity.

```python
from textblob import TextBlob

print("--- Real-Time Sentiment Analyzer ---")
print("Type any sentence to analyze.")
print("Type 'exit' to quit.\n")

while True:
    user_input = input("Enter Text: ")

    if user_input.lower().strip() == "exit":
        print("Exiting Analyzer.")
        break

    if user_input.strip() == "":
        continue

    blob = TextBlob(user_input)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0.1:
        sentiment = "POSITIVE 🙂"
    elif polarity < -0.1:
        sentiment = "NEGATIVE 🙁"
    else:
        sentiment = "NEUTRAL 😐"

    print("-> Sentiment:", sentiment)
    print(f"-> Polarity: {polarity:.2f}")
    print(f"-> Subjectivity: {subjectivity:.2f}\n")
```
