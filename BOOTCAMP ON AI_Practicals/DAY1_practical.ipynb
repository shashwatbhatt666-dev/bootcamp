```python
### Introduction to Python for AI

# Step 1: Introduction to Python
# Python is a versatile programming language widely used in AI and data science.
# Let's start with some basics.

# Print a simple message
print("Welcome SRM Students!")
```

```python
# Step 2: Variables and Data Types
# Variables are used to store data. Python supports multiple data types.

# Example:
x = 5  # Integer
y = 3.14  # Float
name = "AI Enthusiast"  # String
is_learning = True  # Boolean

```

```python
# Print the variables
print("Integer example:", x)
print("Float example:", y)
print("String example:", name)
print("Boolean example:", is_learning)
```

```python
# Step 3: Basic Operations
# Perform arithmetic operations
x=100
y=500
sum_result = x + y
print("Sum of x and y:", sum_result)
```

```python
# Step 4: Lists (to store multiple values)
names = ["Suraj","Pallavi","Sujata","Anita"]
numbers = [1, 2, 3, 4, 5,6,7]
str = "Bootcamp on AI"
print("List of numbers:", numbers)
lst = str.split()
for i in range(len(lst)-1,-1,-1):
    print(lst[i])
```

```python
# Access elements by index
print("First number in the list:", numbers[3])
print(" number in the list:", numbers[5])
```

```python
# Step 5: For Loops
# Loops are used for repetitive tasks
print("Numbers in the list:")
for x in numbers:
    print(x)
```

```python
# Step 6: Functions
# Functions are reusable blocks of code

def greet_user(name):
    print("Hello,", name, "Welcome to AI!")
```

```python
# Call the function
greet_user("Student")
```

```python
# Step 7: Python Libraries
# Python libraries like NumPy, Pandas, and Matplotlib are essential for AI.

# Example: Importing a library and using it
import numpy as np

# Create a simple array
array1 = np.array([1, 2, 3, 4, 5])
print("NumPy Array:", array1)
```

```python
# Perform an operation on the array
squared_array = array1 ** 2
print("Squared Array:", squared_array)
```

```python
# Step 8: Plotting Data
# Visualization is key in AI. Let's use Matplotlib to create a simple plot.
import matplotlib.pyplot as plt

# Example data
years = [2015, 2016, 2017, 2018, 2019]
values = [10, 20, 30, 40, 50]

# Create a line plot
plt.plot(years, values, marker='x')
plt.title("AI Growth Over Years from 2015 to 2019")
plt.xlabel("Year")
plt.ylabel("Value")
plt.show()
```
